# Trying a model of DrBenchmark on a sentence

A run of the benchmark keeps its best model. This notebook loads one of these models and shows what it
predicts for a sentence that you type: the entities are highlighted in the sentence.

It uses the functions of `predict.py`: open it from the root of the repository, in the environment of
the benchmark, with Jupyter installed (`pip install notebook`). A model has to be trained first, e.g.
`python run.py --tasks cas-pos --models Dr-BERT/DrBERT-7GB`.

## The models that the runs kept

The same table as `python predict.py --list`.

```python
import predict

print(predict.list_runs())
```

## Choose the task and the model

A task and a model of the table above. With several runs, the model is the one of the run with the best
score on the validation set: give a seed to take another one.

```python
TASK = "cas-pos"
MODEL = "Dr-BERT/DrBERT-7GB"
SEED = None

model = predict.load(TASK, MODEL, SEED)
print(predict.what_is_loaded(model))
print("The task reads:", ", ".join(model.task.inputs))
```

## Type a sentence

One text for each text that the task reads: one sentence for most tasks, two for a task that compares
sentences, a question and its five answers for FrenchMedMCQA.

```python
from IPython.display import HTML, display

TEXTS = ["Le patient présente une toux sèche depuis trois jours, sans fièvre."]

example = model.example(*TEXTS)
prediction = model.predict([example])[0]
display(HTML(predict.as_html(model, example, prediction)))
```

## Type several sentences

The cell asks for texts until an empty one.

```python
while True:
    texts = [input(f"{column}: ") for column in model.task.inputs]
    if not any(text.strip() for text in texts):
        break
    example = model.example(*texts)
    display(HTML(predict.as_html(model, example, model.predict([example])[0])))
```
